```python
from optix import Var
from optix.simplex import to_tableau, solve
```

```python
x = Var("x")
y = Var("y")
```

```python
objective = 30*x + 40*y
objective
```

```python
constraints = [
    2*x   +   y  <= 10,
      x   +   y  <=  7,
      x   + 2*y  <= 12
]
constraints
```

```python
obj_equality = objective*-1 <= 0
obj_equality
```

```python
tbl, variables = to_tableau(objective, constraints)
variables, tbl
```

```python
solve(objective, constraints)
```
